```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
```

```python
df = pd.read_csv(r"C:\Users\DELL\Downloads\Advertising.csv")
df
```

```python
df.rename({"Unnamed: 0":"a"},axis ="columns",inplace = True)
df
```

```python
df.drop(["a"],axis =1,inplace = True)
df
```

```python
x = df.drop("sales" , axis = 1)
x
```

```python
y = df['sales']
y
```

```python
names = x.columns
names
```

```python
rng = range(len(names))
rng
```

```python
ls = Lasso(alpha = 0.01)
model = ls.fit(x,y)
```

```python
cf = model.coef_
cf
```

```python
plt.plot(rng,cf)
plt.xticks(rng,names,rotation=60)
plt.ylabel("Coefficient")
plt.show()
```

```python
ls.score(x,y)
```

```python
newdata = [[150,50,70]]
sales = ls.predict(newdata)
sales
```
