```python
q3 = df['Weed Waste'].quantile(0.75)
print(q3)
```

```python
q1 = df['Weed Waste'].quantile(0.25)
print(q1)
```

```python
iqr = q3-q1
print(iqr)
```

```python
ul = q3 + (1.5 * iqr)
print(ul)
```

```python
ll = q1 - (1.5 * iqr)
print(ll)
```

```python
outliers_above_UL = df[df['Weed Waste'] > ul]
print(outliers_above_UL)
```

```python
outliers_LL = df[df['Weed Waste'] < ll]
print(outliers_LL)
```

```python
upper = np.where(df['Weed Waste']>=ul)
print(upper)
```

```python
lower = np.where(df['Weed Waste']<=ll)
print(lower)
```
