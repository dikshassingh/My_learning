```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import ElasticNet
```

```python
df = pd.read_csv(r"C:\Users\DELL\Downloads\diabetes.csv")
df
```

```python
x = df.drop("Outcome" , axis = 1)
x
```

```python
y = df['Outcome'].values
y
```

```python
names = x.columns
names
```

```python
rng = range(len(names))
rng
```

```python
es = ElasticNet(l1_ratio=0.5)
model = es.fit(x,y)
es_coef = model.coef_
es.coef_
```

```python
n = np.argmax(es_coef)
n
```

```python
plt.plot(rng, es_coef)
plt.xticks(rng,names,rotation=60)
plt.ylabel("Coefficient")
plt.show()
```
